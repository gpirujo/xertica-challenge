```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path().resolve().parents[1]))
```

```python
import httpx
import json

BASE_URL = "http://api:8000"
```

```python
r = httpx.get(f"{BASE_URL}/health")
print(r.status_code, r.json())
```

```python
r = httpx.get(f"{BASE_URL}/api/v1/alerts/TEST-001/status")
print(r.status_code, r.json())
```

```python
payload = {
    "customer_id": "CUST-001",
    "alert_type": "transferencia_internacional",
    "description": "Transferencia internacional de alto monto desde cuenta corporativa"
}
r = httpx.post(f"{BASE_URL}/api/v1/alerts/TEST-001/analyze", json=payload, timeout=60)
print(r.status_code)
print(json.dumps(r.json(), indent=2, ensure_ascii=False))
```

```python
payload = {
    "customer_id": "CUST-PEP-001",
    "alert_type": "transferencia_internacional",
    "description": "Transferencia desde cuenta de persona políticamente expuesta por $80,000 USD"
}
r = httpx.post(f"{BASE_URL}/api/v1/alerts/TEST-PEP-001/analyze", json=payload, timeout=60)
print(r.status_code)
print(json.dumps(r.json(), indent=2, ensure_ascii=False))
```
